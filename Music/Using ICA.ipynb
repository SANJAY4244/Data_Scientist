```python
import numpy as np
import librosa
from sklearn.decomposition import FastICA
import soundfile as sf

# Load MP3 file
audio, sr = librosa.load("/content/Unnodu_Vaazhatha_K_S_Chitra.mp3", mono=False)

# If stereo, transpose
X = audio.T

# Apply ICA
ica = FastICA(n_components=2, random_state=0)
sources = ica.fit_transform(X)

# Save output files
sf.write("vocals.wav", sources[:,0], sr)
sf.write("music.wav", sources[:,1], sr)
```
